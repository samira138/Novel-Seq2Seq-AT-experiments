```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
file_path = r"h:/Projects/Edit-Novel-Seq2Seq-Module-master/prepare_data/data/AT_data_features.csv"
df = pd.read_csv(file_path)

print("✔ Data Loaded!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

```python
print(" Basic Info:")
df.info()

print("\n Statistical Summary:")
display(df.describe())

print("\n Missing Values:")
print(df.isna().sum())
```

```python
output_dir = "explore_AT_outputs"
os.makedirs(output_dir, exist_ok=True)
print("✔ Output directory created:", output_dir)
```

```python
# هیستوگرام همه فیچرها
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(), bins=50, color='steelblue')
    plt.title(f"Histogram - {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.savefig(f"{output_dir}/hist_{col}.png", dpi=150)
    plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(4, 5))
    plt.boxplot(df[col].dropna(), vert=True)
    plt.title(f"Boxplot - {col}")
    plt.ylabel(col)
    plt.grid(alpha=0.3)
    plt.savefig(f"{output_dir}/boxplot_{col}.png", dpi=150)
    plt.show()
```

```python
#رسم correlation Heatmap 
plt.figure(figsize=(14,10))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Heatmap - AT Dataset")
plt.tight_layout()
plt.savefig(f"{output_dir}/correlation_heatmap.png", dpi=150)
plt.show()
```

```python
sample_df = df.sample(min(2000, len(df)), random_state=42)

sns.pairplot(sample_df[numeric_cols[:]], diag_kind="hist")
plt.savefig(f"{output_dir}/pairplot_sample.png", dpi=150)
plt.show()
```

```python
df.to_csv(f"{output_dir}/AT_clean_preview.csv", index=False)
print("✔ Preview saved.")
```

```python
import pandas as pd
from sklearn.feature_selection import mutual_info_regression
from sklearn.preprocessing import StandardScaler
import numpy as np

# -----------------------------
# 1) انتخاب X و y  
# -----------------------------
# فرض: df دیتافریم اصلی است
y = df['load']
X = df.drop(columns=['load'])

# اگر ستون‌های غیرعددی داری حذف کن
X = X.select_dtypes(include=[np.number])

# -----------------------------
# 2) MI نیاز دارد ورودی‌ها Scale شوند  
# (بهترین حالت: StandardScaler)
# -----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# -----------------------------
# 3) محاسبه Mutual Information
# -----------------------------
mi = mutual_info_regression(X_scaled, y, random_state=42)

# تبدیل به دیتافریم مرتب‌شده
mi_df = pd.DataFrame({
    'Feature': X.columns,
    'MI': mi
}).sort_values(by='MI', ascending=False)

# -----------------------------
# 4) نمایش نتیجه
# -----------------------------
print(mi_df)

# -----------------------------
# 5) رسم نمودار میله‌ای (اختیاری)
# -----------------------------
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.barh(mi_df['Feature'], mi_df['MI'])
plt.title("Mutual Information with Load")
plt.gca().invert_yaxis()
plt.show()
```
